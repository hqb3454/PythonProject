```python
from typing import List
#注意，windows读取文件默认使用gbk，读取utf-8文件会报错，因此需要指定用utf-8读取
def split_into_chunks(doc_file: str) -> List[str]: #->指出返回值类型，原来的整段文本是str，拆分成多端str，然后分别存到不同list中，形成List[str]
    with open(doc_file, 'r',encoding = 'utf-8') as file:#open doc as file只是尝试打开目标路径的文件，并传给file，此时还没有进行文件读取，而read才是真正进行文件读取
        content = file.read()  #with 是上下文管理器，确保读取结束后自动关闭，with 和open不是连在一起的

    return [chunk for chunk in content.split("\n\n")]#split 是字符串自带的切分工具，这里"\n\n"表示按照一个空行进行分割，把原文切分成一个个列表
    #chunk for chunk in，这里for chunk in content.split本身就进行定义了chunk用于接收分好的一个个文本列表，然后依次传入新列表中，新列表为chunk
    

chunks = split_into_chunks("os.md")

for i, chunk in enumerate(chunks):
    print(f"[{i}] {chunk}\n")
```

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer("shibing624/text2vec-base-chinese")

def embed_chunk(chunk: str) -> List[float]:#chunk str 是指某个特定的片段，转化为list float即转化为了对应的向量，之后循环调用该函数即可获取到所有片段对应的向量
    embedding = embedding_model.encode(chunk, normalize_embeddings=True)
    return embedding.tolist()


embedding = embed_chunk("测试内容")
print(len(embedding))
print(embedding)
```

```python
embeddings = [embed_chunk(chunk) for chunk in chunks]
#循环调用，获取到每个片段的向量
print(len(embeddings))
print(embeddings[0])
```

```python
import chromadb
#使用向量数据库
chromadb_client = chromadb.EphemeralClient()
chromadb_collection = chromadb_client.get_or_create_collection(name="default")
#collection相当于数据库的表格，新建表格，然后添加数据
def save_embeddings(chunks: List[str], embeddings: List[List[float]]) -> None:#参数为片段列表和对应的向量列表
    for i, (chunk, embedding) in enumerate(zip(chunks, embeddings)):
        chromadb_collection.add(
            documents=[chunk],
            embeddings=[embedding],
            ids=[str(i)]
        )#依次把片段以及向量传入数据库

save_embeddings(chunks, embeddings)
```

```python
def retrieve(query: str, top_k: int) -> List[str]:
    query_embedding = embed_chunk(query)#把用户问题转化为向量
    results = chromadb_collection.query(
        query_embeddings=[query_embedding],
        n_results=top_k
    )
    return results['documents'][0]

query = "什么是操作系统？"
retrieved_chunks = retrieve(query, 10)

for i, chunk in enumerate(retrieved_chunks):
    print(f"[{i}] {chunk}\n") #召回阶段
```

```python
from sentence_transformers import CrossEncoder

def rerank(query: str, retrieved_chunks: List[str], top_k: int) -> List[str]:
    cross_encoder = CrossEncoder('cross-encoder/mmarco-mMiniLMv2-L12-H384-v1')
    pairs = [(query, chunk) for chunk in retrieved_chunks]
    scores = cross_encoder.predict(pairs)

    scored_chunks = list(zip(retrieved_chunks, scores))
    scored_chunks.sort(key=lambda x: x[1], reverse=True)

    return [chunk for chunk, _ in scored_chunks][:top_k]

reranked_chunks = rerank(query, retrieved_chunks, 3)

for i, chunk in enumerate(reranked_chunks):
    print(f"[{i}] {chunk}\n")
```
